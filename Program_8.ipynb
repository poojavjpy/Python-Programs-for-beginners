### ***Write a Python program to demonstrate Tuple operations.***





```python
# Program to demonstrate Tuple Operations

# Creating tuples
tuple1 = (10, 20, 30, 40, 50)
tuple2 = (60, 70, 80)

print("Tuple 1:", tuple1)
print("Tuple 2:", tuple2)

# 1. Accessing elements
print("\n1. Accessing Tuple Elements")
print("First element of tuple1:", tuple1[0])
print("Last element of tuple1:", tuple1[-1])

# 2. Slicing
print("\n2. Tuple Slicing")
print("Elements from index 1 to 3:", tuple1[1:4])
print("First three elements:", tuple1[:3])
print("Elements from index 2 onwards:", tuple1[2:])

# 3. Concatenation
print("\n3. Tuple Concatenation")
tuple3 = tuple1 + tuple2
print("Tuple1 + Tuple2:", tuple3)

# 4. Repetition
print("\n4. Tuple Repetition")
tuple4 = tuple2 * 2
print("Tuple2 repeated 2 times:", tuple4)

# 5. Membership operators
print("\n5. Membership Operations")
print("30 is present in tuple1:", 30 in tuple1)
print("100 is present in tuple1:", 100 in tuple1)

# 6. Length of tuple
print("\n6. Length of Tuple")
print("Length of tuple1:", len(tuple1))

# 7. Maximum and minimum
print("\n7. Maximum and Minimum Values")
print("Maximum value:", max(tuple1))
print("Minimum value:", min(tuple1))

# 8. Sum of elements
print("\n8. Sum of Tuple Elements")
print("Sum:", sum(tuple1))

# 9. Count operation
tuple5 = (10, 20, 10, 30, 10, 40)
print("\n9. Count Operation")
print("Tuple:", tuple5)
print("Number of times 10 occurs:", tuple5.count(10))

# 10. Index operation
print("\n10. Index Operation")
print("Index of 30:", tuple5.index(30))

# 11. Iterating through tuple
print("\n11. Iterating Through Tuple")
for value in tuple1:
    print(value, end=" ")

# 12. Nested tuple
nested_tuple = ((1, 2), (3, 4), (5, 6))
print("\n\n12. Nested Tuple")
print("Nested Tuple:", nested_tuple)
print("Element from nested tuple:", nested_tuple[1][0])

# 13. Tuple unpacking
print("\n13. Tuple Unpacking")
student = ("Rahul", 21, "Computer Science")
name, age, course = student

print("Name:", name)
print("Age:", age)
print("Course:", course)

# 14. Converting list into tuple
print("\n14. Converting List into Tuple")
my_list = [100, 200, 300]
converted_tuple = tuple(my_list)
print("List:", my_list)
print("Converted Tuple:", converted_tuple)

# 15. Deleting entire tuple
print("\n15. Tuple is Immutable")
print("Individual tuple elements cannot be modified or deleted.")
```

### ***And write a Python program to perform Set operations (union,intersection, difference).***

```python
# Program to demonstrate Set Operations

# Creating two sets
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

print("Set 1:", set1)
print("Set 2:", set2)

# 1. Union
print("\n1. Union Operation")
union_set = set1.union(set2)
print("Union using union():", union_set)

# Union using | operator
print("Union using | operator:", set1 | set2)

# 2. Intersection
print("\n2. Intersection Operation")
intersection_set = set1.intersection(set2)
print("Intersection using intersection():", intersection_set)

# Intersection using & operator
print("Intersection using & operator:", set1 & set2)

# 3. Difference
print("\n3. Difference Operation")

difference1 = set1.difference(set2)
difference2 = set2.difference(set1)

print("Set1 - Set2:", difference1)
print("Set2 - Set1:", difference2)

# Difference using - operator
print("Set1 - Set2 using operator:", set1 - set2)

# 4. Symmetric Difference
print("\n4. Symmetric Difference")
symmetric_diff = set1.symmetric_difference(set2)

print("Symmetric Difference:", symmetric_diff)
print("Using ^ operator:", set1 ^ set2)

# 5. Add an element
print("\n5. Adding an Element")
set1.add(60)
print("Set1 after adding 60:", set1)

# 6. Update multiple elements
print("\n6. Adding Multiple Elements")
set1.update([70, 80, 90])
print("Set1 after update:", set1)

# 7. Remove an element
print("\n7. Removing an Element")
set1.remove(20)
print("Set1 after removing 20:", set1)

# 8. Discard an element
print("\n8. Discard Operation")
set1.discard(100)
print("Set1 after discarding 100:", set1)

# discard does not produce an error even if element is absent

# 9. Membership
print("\n9. Membership Operation")
print("30 is present in set1:", 30 in set1)
print("100 is present in set1:", 100 in set1)

# 10. Length of Set
print("\n10. Length of Set")
print("Number of elements in set1:", len(set1))

# 11. Maximum and Minimum
print("\n11. Maximum and Minimum")
print("Maximum value:", max(set1))
print("Minimum value:", min(set1))

# 12. Sum of elements
print("\n12. Sum of Set Elements")
print("Sum:", sum(set1))

# 13. Subset
print("\n13. Subset Operation")
A = {1, 2}
B = {1, 2, 3, 4}

print("A =", A)
print("B =", B)

print("Is A subset of B?", A.issubset(B))

# 14. Superset
print("\n14. Superset Operation")
print("Is B superset of A?", B.issuperset(A))

# 15. Disjoint sets
print("\n15. Disjoint Set Operation")
C = {5, 6, 7}

print("A =", A)
print("C =", C)

print("Are A and C disjoint?", A.isdisjoint(C))

# 16. Copy a set
print("\n16. Copying a Set")
new_set = set2.copy()

print("Original Set:", set2)
print("Copied Set:", new_set)

# 17. Iterating through a set
print("\n17. Iterating Through Set")

for item in set2:
    print(item, end=" ")

# 18. Remove all elements
print("\n\n18. Clear Operation")

temp_set = {1, 2, 3, 4}
print("Before clear:", temp_set)

temp_set.clear()

print("After clear:", temp_set)
```
