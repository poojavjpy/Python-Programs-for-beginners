# NumPy Mathematical Operations on a Dataset


**Question:** Develop a NumPy solution that uses universal functions and array operations to perform mathematical computations efficiently on a large dataset.



```python
import numpy as np
print("NumPy is ready!")
```

## Program 1 — Basic calculations on an array
An array stores several numbers. Adding or multiplying a number applies the operation to **every element**.

```python
marks = np.array([40, 50, 60, 70, 80])
print("Original marks:", marks)
print("Add 5 marks:", marks + 5)
print("Double marks:", marks * 2)
```

## Program 2 — Calculate 10% of each value
**Mathematics:** 10% of x = x × 10 / 100.

```python
prices = np.array([100, 200, 300, 400, 500])
discount = prices * 10 / 100
print("Prices:", prices)
print("10% discount:", discount)
print("After discount:", prices - discount)
```

## Program 3 — Simple universal functions (ufuncs)
`np.abs()` removes the negative sign; `np.sqrt()` gives square roots; `np.round()` rounds values to the nearest integer (ties to even).

```python
numbers = np.array([-4, -2, 0, 2, 4])
print("Absolute values:", np.abs(numbers))
squares = np.array([4, 9, 16, 25])
print("Square roots:", np.sqrt(squares))
decimals = np.array([2.3, 3.8, 5.1])
print("Rounded values:", np.round(decimals))
```

## Program 4 — Find total, average, maximum and minimum
These operations help summarize a dataset.

```python
sales = np.array([120, 150, 100, 180, 200])
print("Sales:", sales)
print("Total sales:", np.sum(sales))
print("Average sales:", np.mean(sales))
print("Highest sale:", np.max(sales))
print("Lowest sale:", np.min(sales))
```

## Program 5 — A simple shop example
Quantity × price gives the amount for each product. NumPy calculates all amounts in one line.

```python
quantity = np.array([2, 3, 1, 4, 2])
price = np.array([50, 30, 100, 20, 60])
amount = quantity * price
print("Quantity:", quantity)
print("Price:", price)
print("Amount for each product:", amount)
print("Total bill:", np.sum(amount))
```

## Program 6 — The same idea on a larger dataset
`np.arange(1, 1001)` creates numbers from **1 to 1000**. We do not print all 1000 values, only a few results. This shows how one NumPy instruction works on a large array.

```python
data = np.arange(1, 1001)
result = data * 2
print("Number of values:", data.size)
print("First 10 original values:", data[:10])
print("First 10 doubled values:", result[:10])
print("Total of original values:", np.sum(data))
print("Average:", np.mean(data))
```
