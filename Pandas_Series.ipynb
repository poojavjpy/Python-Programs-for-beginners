A. Introduction and Uses of Pandas
1. What is Pandas?
Pandas is an open-source Python library used for data analysis and manipulation. It provides two main data structures: Series and DataFrame.
2. Why do we use Pandas?
Pandas makes it easy to organize, process, filter and analyze data. It also supports reading and writing data in different file formats, such as CSV and Excel.
3. Difference between NumPy and Pandas
NumPy is mainly used for numerical computations and mathematical operations on arrays. Pandas is mainly used for analyzing and manipulating labelled and tabular data.
4. Series and DataFrame as Pandas Data Structures
- Series: A one-dimensional labelled array that stores data along with index labels.
- DataFrame: A two-dimensional labelled data structure that stores data in rows and columns, similar to an Excel sheet.
5. Applications of Pandas
- Data Cleaning: Handling missing values, removing duplicates and correcting incorrect data.
- Data Analysis: Calculating averages, totals and identifying patterns in data.
- Data Manipulation: Adding, deleting, modifying, sorting and filtering data according to requirements.

# Series
### deals with 1D labelled array.


```python
import pandas as pd
a=[1,2,3,4,5]
b=pd.Series(a)
print(b)
print()
print(type(b))
print()
print(b[3])
```

```python
a=[1,2,3,4,5]
b=pd.Series(a,index=['p','o','o','j','a'])
print(b)
print()
print(type(b))
print()
print(b[3])
```

```python
a=[1,2,3,4,5]
b=pd.Series(a,index=['p','o','o','j','a'],dtype="float",name="python Pandas")
print(b)
print()
print(type(b))
print()
print(b[3])
```

Create a Series from a dictionary.

Dictionary keys become index labels and dictionary values become Series values.

```python
# Create a dictionary of subject marks
data = {"Python": 85, "Maths": 75, "English": 80}

# Convert the dictionary into a Series
subjects = pd.Series(data)
print(subjects)
```

For accessing values
Use .loc for labels and .iloc for integer positions (starting from zero).

```python
# Access a value using its label
print("Label p:", b.loc["p"])

# Access the first value using its position
print("First value:", b.iloc[4])

# Access multiple values
print(b.loc[["p", "j"]])
```

#  Modify, add and delete values

```python


# Create a fresh Series
scores = pd.Series([70, 85, 90], index=["A", "B", "C"])
print("Original Series:")
print(scores)

# Modify an existing value
scores.loc["B"] = 88
print("\nAfter Modification:")
print(scores)

# Add a new value with a new label
scores.loc["D"] = 95
print("\nAfter Addition:")
print(scores)

# Delete a value using its label
scores = scores.drop("A")
print("\nAfter Deletion:")
print(scores)
```

# Series attributes and arithmetic

```python
# Display all values of the Series as a NumPy array
print("Values:", scores.to_numpy())  #Converts Series values into a NumPy array without index labels.
print()

# Display the total number of elements
print("Size:", scores.size)
print()

# Display the data type of elements
print("Data type:", scores.dtype)
print()

# Add 5 to every value of the Series
print("After adding 5:")
print(scores + 5)
print()

# Display the original Series
print("Original Series:")
print(scores)
```

### Index Objects
An Index object stores the labels of a Series. Index labels help identify and align data. Individual index labels cannot be modified directly because Index objects are immutable; however, the entire index can be replaced.

**Inspect and create an Index**

```python


# Create a Series with custom index labels
s = pd.Series([100, 200, 300], index=["a", "b", "c"])

# Display the original Series
print("Original Series:")
print(s)
print()

# Display the Index object of the Series
print("Index Object:", s.index)
print()

# Access the first index label
print("First label:", s.index[0])
print()

# Check whether label 'b' exists in the index
print("Is b present?", "b" in s.index)
print()

# Create an independent Index object
idx = pd.Index(["X", "Y", "Z"])

```

Replace index labels---
Replacing labels does not rearrange the values. The new index must have the same length.

```python
# Replace the entire index
s.index = ["X", "Y", "Z"]
print(s)
```

Reindex----
The reindex() method creates a Series with a requested set of labels. Existing values follow their labels. New labels receive missing values (NaN) unless a fill value is supplied. The original Series remains unchanged.

```python
# Create original Series
marks = pd.Series([70, 85, 90], index=["A", "B", "C"])
print(marks)
print()
# Change the order of labels
result = marks.reindex(["C", "A", "B"])
print(result)
```

**Add a missing label and remove a label**

```python
# D is a new label; B is not requested
result = marks.reindex(["A", "C", "D"])
print(result)
print()
# Check which values are missing
print(result.isna())
```

Fill newly introduced missing values
Here, zero is used only as an example. In real data, missing marks do not necessarily mean zero marks.

```python
# Assign 0 to newly introduced labels
result = marks.reindex(["A", "B", "C", "D"], fill_value=0)
print(result)
print()

# Verify that the original Series is unchanged
print("Original:")
print(marks)
```

Drop Entry in Series--
The drop() method removes one or more elements from a Series using their index labels.

```python
import pandas as pd

# Create a Series
s = pd.Series([10, 20, 30, 40],
              index=["a", "b", "c", "d"])

print("Original Series:")
print(s)
print()

# Drop a single element
print("After Dropping b:")
print(s.drop("b"))
print()

# Drop multiple elements
print("After Dropping a and c:")
print(s.drop(["a", "c"]))
```

Selecting Entries in Series--
Series elements can be selected using index labels, integer positions, slicing and conditions.

```python
import pandas as pd

# Create a Series
s = pd.Series([70, 85, 90, 65],
              index=["A", "B", "C", "D"])

print("Original Series:")
print(s)
print()

# Select an element using its label
print("Using loc:")
print(s.loc["B"])
print()

# Select an element using its position
print("Using iloc:")
print(s.iloc[1])
print()

# Select multiple elements
print("Selecting Multiple Elements:")
print(s[["A", "C"]])
print()

# Select first two elements
print("Using Slicing:")
print(s.iloc[0:2])
print()

# Select values greater than 75
print("Using Condition:")
print(s[s > 75])
```

Data Alignment in Series
When arithmetic operations are performed between two Series, Pandas automatically matches their index labels.

image.png

```python
import pandas as pd

# Create two Series with different index labels
s1 = pd.Series([10, 20, 30],
               index=["a", "b", "c"])

s2 = pd.Series([100, 200, 300],
               index=["b", "c", "d"])

print("First Series:")
print(s1)
print()

print("Second Series:")
print(s2)
print()

# Add Series according to matching labels
print("After Addition:")
print(s1 + s2)
print()

# Treat unmatched values as zero
print("Addition using fill_value:")
print(s1.add(s2, fill_value=0))
```

Rank and Sort in Series--
Ranking assigns positions to values, while sorting arranges values or index labels in a particular order.

```python
import pandas as pd

# Create a Series
s = pd.Series([70, 95, 85, 60],
              index=["B", "D", "A", "C"])

print("Original Series:")
print(s)
print()

# Assign rank 1 to the highest value
print("Ranking:")
print(s.rank(ascending=False))
print()

# Sort values in ascending order
print("Ascending Order:")
print(s.sort_values())
print()

# Sort values in descending order
print("Descending Order:")
print(s.sort_values(ascending=False))
print()

# Sort according to index labels
print("Sorting by Index:")
print(s.sort_index())
```

Summary Statistics in Series--
Summary statistics describe the numerical characteristics of a Series.

```python
import pandas as pd

# Create a Series of marks
s = pd.Series([70, 80, 90, 60, 85])

print("Original Series:")
print(s)
print()

# Calculate total
print("Sum:", s.sum())
print()

# Calculate average
print("Mean:", s.mean())
print()

# Find middle value
print("Median:", s.median())
print()

# Find minimum value
print("Minimum:", s.min())
print()

# Find maximum value
print("Maximum:", s.max())
print()

# Count non-missing values
print("Count:", s.count())
print()

# Calculate standard deviation
print("Standard Deviation:", s.std())
print()

# Display statistical summary
print("Summary Statistics:")
print(s.describe())
```

Missing Data in Series--
Missing values in a Series can be detected, removed or replaced using Pandas methods.

```python
import pandas as pd
import numpy as np

# Create a Series with missing values
s = pd.Series([70, np.nan, 90, np.nan, 80])

print("Original Series:")
print(s)
print()

# Identify missing values
print("Missing Values:")
print(s.isna())
print()

# Count missing values
print("Total Missing Values:", s.isna().sum())
print()

# Remove missing values
print("After Removing Missing Values:")
print(s.dropna())
print()

# Replace missing values with zero
print("Replacing Missing Values with Zero:")
print(s.fillna(0))
print()

# Replace missing values with mean
print("Replacing Missing Values with Mean:")
print(s.fillna(s.mean()))
```

Index Hierarchy in Series
Hierarchical indexing allows a Series to contain more than one level of index labels.

```python
import pandas as pd

# Create two levels of index labels
idx = pd.MultiIndex.from_arrays(
    [["BSc", "BSc", "BCA", "BCA"],
     ["Aman", "Riya", "Neha", "Rahul"]],
    names=["Department", "Student"]
)

# Create a Series with hierarchical indexing
s = pd.Series([80, 90, 85, 75], index=idx)

print("Hierarchical Series:")
print(s)
print()

# Select all BSc students
print("BSc Students:")
print(s.loc["BSc"])
print()

# Select marks of a particular student
print("Marks of Riya:")
print(s.loc[("BSc", "Riya")])
print()

# Convert hierarchical Series into DataFrame
print("After Resetting Index:")
print(s.reset_index(name="Marks"))
```
