### ***And write a Python program to perform arithmetic operations using user input.***





```python
num1 = float(input("Enter first number (integer or float): "))
num2 = float(input("Enter second number (integer or float): "))

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)

if num2 != 0:
    print("Division:", num1 / num2)
    print("Floor Division:", num1 // num2)
    print("Modulus / Remainder:", num1 % num2)
else:
    print("Division: Cannot divide by zero")
    print("Floor Division: Cannot divide by zero")
    print("Modulus: Cannot divide by zero")

print("Power:", num1 ** num2)
```
