### ***User defined Functions***

Definition:
A user-defined function is a function created by the programmer to perform a specific task. It is created using the def keyword.
Purpose:  
- Reduces repetition of code.
- Makes programs easier to understand and manage.
- Allows the same code to be reused whenever required.





### Syntax of User-Defined Function

```python
def function_name(parameters):
    statements
    return value
```

- **def** → Keyword used to define a function  
- **function_name** → Name given to the function  
- **parameters** → Inputs passed to the function  
- **statements** → Operations performed inside the function  
- **return** → Returns the result to the caller

2. User defined function for calculating area.

````python
def calculate_area(length, width):
    area = length * width
    return area

room_area = calculate_area(12, 8)
print(room_area)
````

3. Program for bill calculator.

````python
def final_bill(amount, tax_rate, discount):
    tax = amount * tax_rate / 100
    total = amount + tax - discount
    return round(total, 2)

print(final_bill(1200, 5, 100))
print(final_bill(850, 12, 50))
````

Check Even or Odd

````python
def check_even_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter a number: "))
print(check_even_odd(num))
````

Maximum of Three Numbers

````python
def maximum(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

x = int(input("Enter first number: "))
y = int(input("Enter second number: "))
z = int(input("Enter third number: "))

print("Maximum =", maximum(x, y, z))
````

Factorial of a Number

````python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact

num = int(input("Enter a number: "))
print("Factorial =", factorial(num))
````

Reverse a Number

````python
def reverse_number(n):
    rev = 0

    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10

    return rev

num = int(input("Enter a number: "))
print("Reverse =", reverse_number(num))
````

Check Palindrome Number

````python
def palindrome(n):
    original = n
    rev = 0

    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10

    if original == rev:
        return "Palindrome"
    else:
        return "Not Palindrome"

num = int(input("Enter a number: "))
print(palindrome(num))
````

Check Prime Number

````python
def check_prime(n):
    if n < 2:
        return "Not Prime"

    for i in range(2, n):
        if n % i == 0:
            return "Not Prime"

    return "Prime"

num = int(input("Enter a number: "))
print(check_prime(num))
````

Fibonacci Series

````python
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        c = a + b
        a = b
        b = c

num = int(input("How many terms? "))
fibonacci(num)
````

Sum of Digits

````python
def sum_digits(n):
    total = 0

    while n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10

    return total

num = int(input("Enter a number: "))
print("Sum of digits =", sum_digits(num))
````

Count Digits in a Number

````python
def count_digits(n):
    count = 0

    while n > 0:
        count = count + 1
        n = n // 10

    return count

num = int(input("Enter a number: "))
print("Number of digits =", count_digits(num))
````

Multiplication Table Using Function

````python
def table(n):
    for i in range(1, 11):
        print(n, "x", i, "=", n * i)

num = int(input("Enter a number: "))
table(num)
````
