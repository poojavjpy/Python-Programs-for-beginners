### ***if you want to print numbers from 1 to 200***

```python
def creater():
    list = []
    i = 1
    while i <= 200:
        list.append(i)
        i += 1
    return list

print(creater())
```

```python
numbers = list(range(1, 201))
print(numbers)
```

### ***lets check the size***

```python
import sys

def creater():
    list = []
    i = 1
    while i <= 200:
        list.append(i)
        i += 1
    return list

print(creater())

z = sys.getsizeof(list)
print("memory size = ", z)
```

### ***lets add each num by 10***

```python
import sys

def creater():
    list = []
    i = 1
    while i <= 200:
        list.append(i)
        i += 1
    return list

print(creater())

z = sys.getsizeof(list)
print("memory size = ", z)
print([num + 10 for num in creater()])  #List Comprehension
print("memory size = ", z)

```

### ***to use the memory efficiently we are not creating list elements..we are generating numbers as per demand***

```python
def creater1():
    i = 1
    while i <= 200:
        yield i
        i += 1

print(creater1())
```

```python
g = creater1()

print(next(g))
#print(next(g))
#print(next(g))
```

### ***we can generate whole list at a time but it remembers its last and current position.***

```python
g = creater1()

print(next(g))
#print(next(g))
#print(next(g))
print(list(g))
```

```python
dir(g)
```

```python
print('__next__' in dir(g))
```

```python
def count_up():
    yield 1
    yield 2
    yield 3

g = count_up()

print(next(g))  # 1
print(next(g))  # 2
```

### ***Memories comparision***

```python
import sys

l = [i for i in range(100000)]

print(sys.getsizeof(l))
```

```python
import sys

# List - stores all elements
l = [i for i in range(100000)]

# Range - generates values when needed
r = range(100000)

# Generator - produces values one by one
g = (i for i in range(100000))

print("List:", sys.getsizeof(l), "bytes")
print("Range:", sys.getsizeof(r), "bytes")
print("Generator:", sys.getsizeof(g), "bytes")
```
