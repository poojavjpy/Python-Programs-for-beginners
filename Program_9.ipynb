### ***Write a Python program to create a Dictionary and perform add, update, delete and search operations.***

```python
# Program to create a Dictionary and perform
# Add, Update, Delete and Search operations

# Creating a dictionary
student = {
    "roll_no": 101,
    "name": "Rahul",
    "course": "BCA",
    "marks": 85
}

print("Original Dictionary:")
print(student)

# 1. Add Operation
print("\n1. Add Operation")

student["city"] = "Delhi"
print("After adding city:")
print(student)

# 2. Update Operation
print("\n2. Update Operation")

student["marks"] = 92
student["course"] = "MCA"

print("After updating marks and course:")
print(student)

# 3. Delete Operation
print("\n3. Delete Operation")

del student["city"]

print("After deleting city:")
print(student)

# Another way to delete using pop()
removed_value = student.pop("course")

print("\nRemoved course:", removed_value)
print("Dictionary after pop():")
print(student)

# 4. Search Operation
print("\n4. Search Operation")

key = input("Enter the key to search: ")

if key in student:
    print("Key found.")
    print("Value =", student[key])
else:
    print("Key not found.")

# 5. Display all keys
print("\n5. Dictionary Keys:")
print(student.keys())

# 6. Display all values
print("\n6. Dictionary Values:")
print(student.values())

# 7. Display all key-value pairs
print("\n7. Dictionary Items:")

for key, value in student.items():
    print(key, ":", value)

# 8. Length of dictionary
print("\n8. Number of elements in dictionary:")
print(len(student))
```
