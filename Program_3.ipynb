### **Write a Python program to check whether a number is even/odd and positive/negative using if-else.**

```python
num = int(input("Enter a number: "))

# Check positive or negative
if num > 0:
    print("The number is Positive")
elif num < 0:
    print("The number is Negative")
else:
    print("The number is Zero")

# Check even or odd
if num % 2 == 0:
    print("The number is Even")
else:
    print("The number is Odd")
```

### ***And write a Python program to print multiplication tables using loops.***

```python
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```
