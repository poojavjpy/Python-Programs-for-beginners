### ***Write a Python program to check whether a string is a palindrome.***

```python
text = input("Enter a string: ")

reverse = text[::-1]

if text == reverse:
    print("The string is a Palindrome")
else:
    print("The string is not a Palindrome")
```

### ***write a Python program to perform basic string operations (length, concatenation, slicing,uppercase, lowercase).***


```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

# Length
print("Length of first string =", len(str1))

# Concatenation
print("Concatenation =", str1 + " " + str2)

# Slicing
print("First 3 characters =", str1[:3])

# Uppercase
print("Uppercase =", str1.upper())

# Lowercase
print("Lowercase =", str1.lower())
```

```python
text = input("Enter a string: ")

# Remove extra spaces
print("Strip =", text.strip())

# Replace text
print("Replace =", text.replace("a", "@"))

# Count characters
print("Count of 'a' =", text.count("a"))

# Find position
print("Position of 'a' =", text.find("a"))

# Capitalize first letter
print("Capitalize =", text.capitalize())

# Convert to title case
print("Title Case =", text.title())

# Check whether string contains only alphabets
print("Is Alphabetic =", text.isalpha())

# Check whether string contains only digits
print("Is Numeric =", text.isdigit())

# Check whether string contains letters and numbers
print("Is Alphanumeric =", text.isalnum())

# Reverse string
print("Reverse =", text[::-1])
```

### **Write a Python program to count vowels, consonants, digits and spaces in a string.**

```python
text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1

print("Vowels =", vowels)
print("Consonants =", consonants)
print("Digits =", digits)
print("Spaces =", spaces)
```
