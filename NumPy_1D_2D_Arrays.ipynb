# Q1 — Creating and Manipulating 1D and 2D Arrays

Demonstrate the creation and manipulation of one-dimensional and two-dimensional NumPy arrays. Illustrate indexing and basic array operations with suitable examples.



```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. Create one-dimensional arrays

A 1D array is a sequence of elements. Compare `array`, `arange`, and `linspace`.

```python
a = np.array([10, 20, 30, 40, 50])
b = np.arange(2, 12, 2)
c = np.linspace(0, 1, 5)
print("a:", a, "shape:", a.shape, "dtype:", a.dtype)
print("b:", b)
print("c:", c)
```

## 2. Create two-dimensional arrays

A 2D array has rows and columns. `shape` reports its dimensions.

```python
marks = np.array([[78, 85, 90], [88, 76, 95], [91, 89, 84]])
print("Marks:\n", marks)
print("Shape:", marks.shape, "Dimensions:", marks.ndim, "Size:", marks.size)
```

## 3. Index and slice 1D arrays

Python indexing begins at zero; slice end indices are excluded.

```python
print("First:", a[0], "Last:", a[-1])
print("Middle values:", a[1:4])
print("Every second value:", a[::2])
```

## 4. Index and slice 2D arrays

Use `[row, column]` to select a cell.

```python
print("Row 2:", marks[1, :])
print("Column 3:", marks[:, 2])
print("Student 1, subject 2:", marks[0, 1])
print("Top-left 2x2:\n", marks[:2, :2])
```

## 5. Manipulate arrays

Update values, reshape, and concatenate arrays. A copy avoids changing the original.

```python
edited = marks.copy()
edited[0, 0] = 82
print("Updated:\n", edited)
print("Reshaped 1D to 2D:\n", a[:4].reshape(2, 2))
print("Concatenated:", np.concatenate([a, b]))
```

## 6. Basic array operations

Arithmetic with a scalar broadcasts to all elements; equal-shaped arrays operate elementwise.

```python
x = np.array([2, 4, 6])
y = np.array([1, 3, 5])
print("Addition:", x + y)
print("Multiplication:", x * y)
print("Squared:", x ** 2)
print("Marks + 5:\n", marks + 5)
print("Subject averages:", marks.mean(axis=0))
```
