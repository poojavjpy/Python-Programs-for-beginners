### ***1. lambda function***





 1. A lambda function is not a completely different kind of programming concept. It's mainly a shorter way of writing a simple function.
 2. A lambda function is a anonymous, small function written in one line.
 3. A lambda can take one, two, three, or more arguments, but its body is a single expression.
 4. The important point is that lambda does not need def or return:
 5. lambda is best for short/simple calculations. For a function with several steps, the normal def version is often easier to read.

### **Syntax**

lambda x : x * x

Before colon (:) → Argument
After colon (:) → Operation / Expression

```python
def greet(name):
  return "Hello, "  + name

print(greet ("Bsc Section 9"))
```

```python
greet = lambda name: "Hello, " + name

print(greet("Asha"))
```

### ***User defined function for calculating area.***

```python
def calculate_area(length, width):
    area = length * width
    return area

room_area = calculate_area(12, 8)
print(room_area)
```

### ***Using lambda, the same function can be written as:***

```python
calculate_area = lambda length, width: length * width

room_area = calculate_area(12, 8)
print(room_area)
```

### ***Program for bill calculator.***

```python
def final_bill(amount, tax_rate, discount):
    tax = amount * tax_rate / 100
    total = amount + tax - discount
    return round(total, 2)

print(final_bill(1200, 5, 100))
print(final_bill(850, 12, 50))
```

### ***Using lambda, the same function can be written as:***

```python
final_bill = lambda amount, tax_rate, discount: round(
    amount + (amount * tax_rate / 100) - discount, 2
)

print(final_bill(1200, 5, 100))
print(final_bill(850, 12, 50))
```

### ***2. map() Function***

map() is used when we want to perform the same operation on every item of a sequence such as a list.

Syntax

map(function, iterable)

Here:
function → operation that will be performed

iterable → collection of values, such as a list

Usually we convert the result into a list

MAP()

Purpose:
map() is used to apply the same function or operation to every item of an iterable.

Syntax:
map(function, iterable)

To get the result as a list:
list(map(function, iterable))

Example:
numbers = [1, 2, 3, 4]
result = list(map(lambda x: x * 2, numbers))

Output:
[2, 4, 6, 8]

Remember:
map() → modifies/transforms every item

### Example 1: Square Every ***Number***

```python
def square(n):
    return n * n

numbers = [1, 2, 3, 4, 5]

result = list(map(square, numbers))

print(result)
```

### ***Example 2: Double Every Number***

```python
numbers = [2, 4, 6, 8]

result = list(map(lambda n: n * 2, numbers))

print(result)
```

### ***Example 3: Add 10 to Every Number***

```python
numbers = [5, 10, 15, 20]

result = list(map(lambda n: n + 10, numbers))

print(result)
```

### ***Example 4: Convert Names to Uppercase***

```python
names = ["asha", "rahul", "priya"]

result = list(map(lambda name: name.upper(), names))

print(result)
```

### ***Example 5: Find Length of Each Word***

```python
words = ["cat", "python", "apple", "computer"]

result = list(map(lambda word: len(word), words))

print(result)
```

### 3. filter() ***Function***



filter() is used to select only those items from a list that satisfy a condition.

### Syntax

filter(function, iterable)

list(filter(function, iterable))

The function used with filter() should give True or False.
- True → keep the item
- False → remove/ignore the item

Example 1: Find Even Numbers

FILTER()

Purpose:
filter() is used to select elements from an iterable based on a condition.

Syntax:
filter(function, iterable)

To get the result as a list:
list(filter(function, iterable))

The condition returns:
True  → Item is selected
False → Item is ignored

Example:
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers))

Output:
[2, 4, 6]

Remember:
filter() → selects items based on a condition

```python
def is_even(n):
    return n % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(is_even, numbers))

print(result)
```

by using lambda

```python
numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda n: n % 2 == 0, numbers))

print(result)
```

Example 2: Find Odd Numbers

```python
numbers = [1, 2, 3, 4, 5, 6, 7]

result = list(filter(lambda n: n % 2 != 0, numbers))

print(result)
```

Example 3: Numbers Greater Than 10

```python
numbers = [5, 12, 3, 18, 7, 20]

result = list(filter(lambda n: n > 10, numbers))

print(result)
```

Example 4: Select Positive Numbers

```python
numbers = [-5, 10, -3, 8, 0, -2, 7]

result = list(filter(lambda n: n > 0, numbers))

print(result)
```

Example 5: Select Words Longer Than 5 Characters

```python
words = ["cat", "python", "apple", "computer", "dog"]

result = list(filter(lambda word: len(word) > 5, words))

print(result)
```

4. REDUCE()

Purpose:
reduce() is used to combine all elements of an iterable and produce a single final value.

Import:
from functools import reduce

Syntax:
reduce(function, iterable)

Example:
numbers = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, numbers)

Output:
10

Working:
1 + 2 = 3
3 + 3 = 6
6 + 4 = 10

Remember:
reduce() → combines many values into one final value

Unlike map() and filter(), we need to import reduce():

from functools import reduce

Example 1: Sum of All Numbers

```python
from functools import reduce

def add(a, b):
    return a + b

numbers = [1, 2, 3, 4, 5]

result = reduce(add, numbers)

print(result)
```

Numbers = [1, 2, 3, 4, 5]

Step 1:
a = 1, b = 2
1 + 2 = 3

Step 2:
a = 3, b = 3
3 + 3 = 6

Step 3:
a = 6, b = 4
6 + 4 = 10

Step 4:
a = 10, b = 5
10 + 5 = 15

Final result = 15

Example 2: Product of All Numbers

```python
from functools import reduce

numbers = [2, 3, 4, 5]

result = reduce(lambda a, b: a * b, numbers)

print(result)
```

Example 3: Find Maximum Number

```python
from functools import reduce

numbers = [10, 25, 8, 40, 15]

result = reduce(lambda a, b: a if a > b else b, numbers)

print(result)
```

Example 4: Find Minimum Number

```python
from functools import reduce

numbers = [10, 25, 8, 40, 15]

result = reduce(lambda a, b: a if a < b else b, numbers)

print(result)
```

Example 5: Join Words Together

```python
from functools import reduce

words = ["I", "love", "Python"]

result = reduce(lambda a, b: a + " " + b, words)

print(result)
```

5. sorted() Function

sorted() is used to arrange the elements of an iterable in a particular order.
By default, it sorts in ascending order.


sorted() → arranges items in order

### **Syntax**



sorted(iterable, key=None, reverse=False)

- iterable → list, tuple, etc. that we want to sort
- key → tells Python what basis should be used for sorting
- reverse=False → ascending order by default
- reverse=True → descending order

Example 1: Numbers in Ascending Order

```python
numbers = [40, 10, 30, 20, 50]

result = sorted(numbers)

print(result)
```

Example 2: Numbers in Descending Order

```python
numbers = [40, 10, 30, 20, 50]

result = sorted(numbers, reverse=True)

print(result)
```

Example 3: Sort Names Alphabetically

```python
names = ["Rahul", "Asha", "Priya", "Karan"]

result = sorted(names)

print(result)
```

Example 4: Sort Words According to Length

```python
words = ["apple", "cat", "elephant", "dog"]

result = sorted(words, key=lambda x: len(x))

print(result)
```

Example 5: Sort Students According to Marks

```python
students = [("Asha", 85), ("Rahul", 72), ("Priya", 91)]

result = sorted(students, key=lambda x: x[1])

print(result)
```

Very important: sort() vs sorted()

sorted() returns a new sorted list. The original list remains unchanged.

sort() changes the original list itself.

```python
numbers = [30, 10, 20]

result = sorted(numbers)

print(result)
print(numbers)
numbers.sort()
print(numbers)
```
