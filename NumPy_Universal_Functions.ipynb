# NumPy Universal Functions

Implement NumPy universal functions to perform mathematical computations on array elements. Show the output for suitable input arrays.



```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 1. What is a ufunc?

A universal function applies a numerical operation elementwise to arrays.

```python
x = np.array([1, 4, 9, 16, 25])
print("Input:", x)
print("Square roots:", np.sqrt(x))
print("Squares:", np.square(x))
```

## 2. Arithmetic ufuncs

Use `add`, `subtract`, `multiply`, `divide`, and `power`.

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])
print("add:", np.add(a, b))
print("subtract:", np.subtract(a, b))
print("multiply:", np.multiply(a, b))
print("divide:", np.divide(a, b))
print("power:", np.power(b, 2))
```

## 3. Rounding, absolute value and comparison

These ufuncs also work elementwise.

```python
data = np.array([-3.7, -1.2, 0., 2.4, 5.9])
print("absolute:", np.abs(data))
print("ceil:", np.ceil(data))
print("floor:", np.floor(data))
print("round:", np.round(data))
print("greater than 0:", np.greater(data, 0))
```

## 6. Practical program: process sensor readings

Convert Celsius to Fahrenheit and calculate temperature differences.

```python
celsius = np.array([18., 21.5, 25., 29., 32.])
fahrenheit = np.add(np.multiply(celsius, 9 / 5), 32)
changes = np.diff(celsius)
print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
print("Successive changes:", changes)
print("Mean Celsius:", np.mean(celsius))
```
