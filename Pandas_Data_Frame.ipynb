A. Introduction and Uses of Pandas
1. What is Pandas?
Pandas is an open-source Python library used for data analysis and manipulation. It provides two main data structures: Series and DataFrame.
2. Why do we use Pandas?
Pandas makes it easy to organize, process, filter and analyze data. It also supports reading and writing data in different file formats, such as CSV and Excel.
3. Difference between NumPy and Pandas
NumPy is mainly used for numerical computations and mathematical operations on arrays. Pandas is mainly used for analyzing and manipulating labelled and tabular data.
4. Series and DataFrame as Pandas Data Structures
- Series: A one-dimensional labelled array that stores data along with index labels.
- DataFrame: A two-dimensional labelled data structure that stores data in rows and columns, similar to an Excel sheet.
5. Applications of Pandas
- Data Cleaning: Handling missing values, removing duplicates and correcting incorrect data.
- Data Analysis: Calculating averages, totals and identifying patterns in data.
- Data Manipulation: Adding, deleting, modifying, sorting and filtering data according to requirements.

Drop Entry in DataFrame
The drop() method is used to remove rows or columns from a DataFrame using their labels.
- axis=0 represents rows.
- axis=1 represents columns.

```python
import pandas as pd

# Create a DataFrame
d = {"A": [10, 20, 30, 40],
     "B": [50, 60, 70, 80],
     "C": [11, 22, 33, 44]}

b = pd.DataFrame(d)

# Display original DataFrame
print("Original DataFrame:")
print(b)
print()

# Delete row with index label 1
print("After dropping row 1:")
print(b.drop(1, axis=0))
print()

# Delete column B
print("After dropping column B:")
print(b.drop("B", axis=1))
```

Note: The original DataFrame remains unchanged unless the result is reassigned or inplace=True is used.

Selecting Entries in DataFrame
Selecting entries means accessing specific rows, columns or individual values from a DataFrame.
- loc[] selects data using labels.
- iloc[] selects data using integer positions.

```python
import pandas as pd

# Create a DataFrame
d = {"Name": ["Aman", "Riya", "Neha", "Rahul"],
     "Marks": [70, 85, 90, 65],
     "Age": [20, 21, 19, 22]}

b = pd.DataFrame(d, index=["a", "b", "c", "d"])

print("Original DataFrame:")
print(b)
print()

# Select a single column
print("Selecting Marks column:")
print(b["Marks"])
print()

# Select multiple columns
print("Selecting Name and Marks:")
print(b[["Name", "Marks"]])
print()

# Select a row using its label
print("Selecting row b:")
print(b.loc["b"])
print()

# Select a row using its position
print("Selecting first row:")
print(b.iloc[0])
print()

# Select a particular value
print("Marks of Riya:")
print(b.loc["b", "Marks"])
print()

# Select students having marks greater than 75
print("Students scoring above 75:")
print(b[b["Marks"] > 75])
```

Data Alignment in DataFrame
Data alignment is the automatic matching of row and column labels when performing operations between two DataFrames. If a matching label is not available, Pandas produces NaN.

image.png

```python
import pandas as pd

# Create first DataFrame
df1 = pd.DataFrame({"A": [10, 20, 30],
                    "B": [40, 50, 60]},
                   index=["x", "y", "z"])

# Create second DataFrame
df2 = pd.DataFrame({"A": [1, 2, 3],
                    "B": [4, 5, 6]},
                   index=["y", "z", "w"])

print("First DataFrame:")
print(df1)
print()

print("Second DataFrame:")
print(df2)
print()

# Add DataFrames using automatic label alignment
print("After Addition:")
print(df1 + df2)
print()

# Treat unmatched values as zero
print("Addition using fill_value:")
print(df1.add(df2, fill_value=0))
```

Rank and Sort in DataFrame
Ranking---
Ranking assigns a numerical position to values according to their order.

```python
import pandas as pd

# Create a DataFrame
d = {"Name": ["Aman", "Riya", "Neha", "Rahul"],
     "Marks": [70, 95, 85, 60]}

b = pd.DataFrame(d)

print("Original DataFrame:")
print(b)
print()

# Assign rank 1 to the highest marks
b["Rank"] = b["Marks"].rank(ascending=False)

print("DataFrame with Rank:")
print(b)
```

Sorting--
Sorting arranges data in ascending or descending order based on values or index labels.

```python
# Sort marks in ascending order
print("Ascending Order:")
print(b.sort_values(by="Marks"))
print()

# Sort marks in descending order
print("Descending Order:")
print(b.sort_values(by="Marks", ascending=False))
print()

# Sort according to index in descending order
print("Sorting by Index:")
print(b.sort_index(ascending=False))
```

Summary Statistics in DataFrame---
Summary statistics provide useful information about data, such as total, average, minimum, maximum and standard deviation.

```python
import pandas as pd

# Create a DataFrame containing subject marks
d = {"Maths": [70, 80, 90, 60],
     "Python": [85, 75, 95, 65]}

b = pd.DataFrame(d)

print("Original DataFrame:")
print(b)
print()

# Calculate total of each column
print("Sum:")
print(b.sum())
print()

# Calculate average of each column
print("Mean:")
print(b.mean())
print()

# Find minimum value
print("Minimum:")
print(b.min())
print()

# Find maximum value
print("Maximum:")
print(b.max())
print()

# Calculate median
print("Median:")
print(b.median())
print()

# Calculate standard deviation
print("Standard Deviation:")
print(b.std())
print()

# Display complete statistical summary
print("Summary Statistics:")
print(b.describe())
```

Missing Data in DataFrame
Missing data refers to values that are unavailable in a dataset. Pandas commonly represents missing numerical values using NaN (Not a Number).


image.png

```python
import pandas as pd
import numpy as np

# Create a DataFrame with missing values
d = {"Name": ["Aman", "Riya", "Neha", "Rahul"],
     "Marks": [80, np.nan, 90, 70],
     "Age": [20, 21, np.nan, 22]}

b = pd.DataFrame(d)

print("Original DataFrame:")
print(b)
print()

# Identify missing values
print("Missing Values:")
print(b.isna())
print()

# Count missing values in each column
print("Number of Missing Values:")
print(b.isna().sum())
print()

# Remove rows containing missing values
print("After Removing Missing Values:")
print(b.dropna())
print()

# Replace missing values with zero
print("After Filling Missing Values:")
print(b.fillna(0))
print()

# Replace missing marks with average marks
print("Replacing Missing Marks with Mean:")
print(b["Marks"].fillna(b["Marks"].mean()))
```

Index Hierarchy in DataFrame--
Index hierarchy, also known as MultiIndex, allows a DataFrame to have multiple levels of index labels. It is useful for organizing grouped data.

image.png

```python
import pandas as pd

# Create a DataFrame
d = {"Department": ["BSc", "BSc", "BCA", "BCA"],
     "Student": ["Aman", "Riya", "Neha", "Rahul"],
     "Marks": [80, 90, 85, 75]}

b = pd.DataFrame(d)

print("Original DataFrame:")
print(b)
print()

# Create hierarchical index using two columns
b = b.set_index(["Department", "Student"])

print("Hierarchical DataFrame:")
print(b)
print()

# Select all BSc students
print("BSc Students:")
print(b.loc["BSc"])
print()

# Select a particular student
print("Marks of Riya:")
print(b.loc[("BSc", "Riya"), "Marks"])
print()

# Convert index labels back into columns
print("After Resetting Index:")
print(b.reset_index())
```
