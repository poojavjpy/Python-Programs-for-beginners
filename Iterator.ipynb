An iterator in Python is an object that retrieves elements one at a time using the next() function and keeps track of its current position.

```python
numbers = [10, 20, 30, 40]

my_iterator = iter(numbers)

print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
```

### **What happens when elements are finished?**

### **Python raises a StopIteration exception because no elements are left.**

```python
numbers = [10, 20]

x = iter(numbers)

print(next(x))
print(next(x))
print(next(x))
```

### Converting an Iterable into an **Iterator**

```python
numbers = [10, 20, 30]

# Iterable
print(numbers)

# Convert iterable into iterator
x = iter(numbers)

# Access elements one by one
print(next(x))
print(next(x))
print(next(x))
```

### ***Memories comparision***

```python
import sys

l = [i for i in range(100000)]

print(sys.getsizeof(l))
```

```python
import sys

# List comprehension
l = [i for i in range(100000)]

# Generator expression
g = (i for i in range(100000))

print("List memory:", sys.getsizeof(l), "bytes")
print("Generator memory:", sys.getsizeof(g), "bytes")
```

```python
import sys

# List - stores all elements
l = [i for i in range(100000)]

# Range - generates values when needed
r = range(100000)

# Generator - produces values one by one
g = (i for i in range(100000))

print("List:", sys.getsizeof(l), "bytes")
print("Range:", sys.getsizeof(r), "bytes")
print("Generator:", sys.getsizeof(g), "bytes")
```

```python
def count_up():
    yield 1
    yield 2
    yield 3

g = count_up()

print(next(g))  # 1
print(next(g))  # 2
```
