# NumPy 2D Array Data Processing (Google Colab Workbook)
Design a NumPy-based solution to create, reshape, index, and manipulate multidimensional arrays for a given data-processing problem.



## Program 1 — Import NumPy and create a 2D marks array
Rows represent students; columns represent subjects (Python, DBMS, AI, Maths).

```python
import numpy as np
subjects = ["Python", "DBMS", "AI", "Maths"]
marks = np.array([
    [78, 82, 91, 85],
    [88, 75, 80, 90],
    [95, 89, 92, 94],
    [72, 80, 78, 83],
    [90, 93, 87, 91]
])
print("Marks (5 students × 4 subjects):\n", marks)
print("Dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Total elements:", marks.size)
print("Data type:", marks.dtype)
```

## Program 2 — Create and reshape a 2D array
Reshaping changes the arrangement, **not** the number of elements. 3 × 4 = 4 × 3 = 12.

```python
numbers = np.arange(1, 13)
print("Original 1D array:", numbers)
array_2d = numbers.reshape(3, 4)
print("Reshaped 3 × 4:\n", array_2d)
print("Reshaped 4 × 3:\n", array_2d.reshape(4, 3))
assert array_2d.ndim == 2
```

## Program 3 — Indexing in a 2D array
Syntax: `array[row, column]`. Python indices begin at 0.

```python
print("First student's all marks:", marks[0, :])
print("Third student's AI marks:", marks[2, 2])
print("All students' Python marks:", marks[:, 0])
print("Last student's last subject:", marks[-1, -1])
```

## Program 4 — Slicing rows and columns
`start:stop` includes start and excludes stop.

```python
print("First 3 students, all subjects:\n", marks[:3, :])
print("All students, first 2 subjects:\n", marks[:, :2])
print("Students 2–4, subjects 2–3:\n", marks[1:4, 1:3])
```

## Program 5 — Manipulate a 2D array safely
Copy the original array before changing scores. Add 5 grace marks to Python, capped at 100.

```python
updated = marks.copy()
updated[:, 0] = np.minimum(updated[:, 0] + 5, 100)
print("Original Python marks:", marks[:, 0])
print("Updated Python marks:", updated[:, 0])
print("Updated full 2D array:\n", updated)
assert np.array_equal(marks[:, 0], [78, 88, 95, 72, 90])
```

## Program 6 — Row-wise and column-wise processing
`axis=1` computes one result per **student** (across columns); `axis=0` computes one result per **subject** (down rows).

```python
totals = marks.sum(axis=1)
averages = marks.mean(axis=1)
subject_averages = marks.mean(axis=0)
print("Student totals:", totals)
print("Student averages:", np.round(averages, 2))
for subject, avg in zip(subjects, subject_averages):
    print(f"{subject} average: {avg:.2f}")
print("Highest student total:", totals.max())
print("Top student (1-based):", np.argmax(totals) + 1)
```

## Program 7 — Transpose and flatten a 2D array
Transpose switches rows and columns. Flatten converts a 2D array to 1D.

```python
print("Original shape:", marks.shape)
print("Transpose shape:", marks.T.shape)
print("Transposed marks:\n", marks.T)
print("Flattened 1D marks:", marks.flatten())
```

## Program 8 — Boolean indexing and a complete report
Find scores at least 90 and students with an average of at least 85.

```python
print("Scores >= 90:", marks[marks >= 90])
print("Count of scores >= 90:", np.count_nonzero(marks >= 90))
print("Students with average >= 85 (1-based):", np.where(averages >= 85)[0] + 1)
print("\nSTUDENT REPORT")
for i, (total, avg) in enumerate(zip(totals, averages), start=1):
    print(f"Student {i}: Total = {total}, Average = {avg:.2f}")
```
