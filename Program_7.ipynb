### ***Write a Python program to perform operations on Lists (append, insert, remove, sort, reverse).***


```python
numbers = [10, 30, 20, 50, 40]

print("Original List:", numbers)

# Append
numbers.append(60)
print("After Append:", numbers)

# Insert
numbers.insert(2, 25)
print("After Insert:", numbers)

# Remove
numbers.remove(30)
print("After Remove:", numbers)

# Sort
numbers.sort()
print("After Sort:", numbers)

# Reverse
numbers.reverse()
print("After Reverse:", numbers)
```

### ***Input from user.***

```python
numbers = list(map(int, input("Enter list elements separated by spaces: ").split()))

print("Original List:", numbers)

# Append
value = int(input("Enter value to append: "))
numbers.append(value)
print("After Append:", numbers)

# Insert
position = int(input("Enter position to insert: "))
value = int(input("Enter value to insert: "))
numbers.insert(position, value)
print("After Insert:", numbers)

# Remove
value = int(input("Enter value to remove: "))
if value in numbers:
    numbers.remove(value)
    print("After Remove:", numbers)
else:
    print("Value not found in list")

# Sort
numbers.sort()
print("After Sort:", numbers)

# Reverse
numbers.reverse()
print("After Reverse:", numbers)
```

### ***And write a Python program to find the largest, smallest and sum of elements in a list.***

```python
numbers = [10, 25, 5, 40, 15]

largest = max(numbers)
smallest = min(numbers)
total = sum(numbers)

print("List:", numbers)
print("Largest element =", largest)
print("Smallest element =", smallest)
print("Sum of elements =", total)
```

### ***Input from user.***

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

print("Largest =", max(numbers))
print("Smallest =", min(numbers))
print("Sum =", sum(numbers))
```
