```python
import numpy as np

# 1. Create an array
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)

# 2. Access an element
print("Element at index 2:", arr[2])

# 3. Modify an element
arr[1] = 25
print("After Modification:", arr)

# 4. Insert a new element
arr = np.insert(arr, 2, 35)
print("After Insertion:", arr)

# 5. Delete an element
arr = np.delete(arr, 3)
print("After Deletion:", arr)

# 6. Add an element at the end
arr = np.append(arr, 60)
print("After Adding:", arr)
```
